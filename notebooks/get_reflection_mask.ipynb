```python
import numpy as np
from PIL import Image

image_path = (
    "data/shadow-gen/stepho_renderings/reflections_4/train/mercedes-reflection_0069.jpg"
)
image = Image.open(image_path)
im_arr = np.array(image)
image.resize((512, 512))
```

```python
im_arr = im_arr[:, :, 0]  # Convert to grayscale

Image.fromarray((im_arr > 200).astype(np.uint8) * 255, mode="L").resize((512, 512))
```

```python
import numpy as np

bg_mask = (np.array(image) > 200).astype(np.uint8)[..., 0]
Image.fromarray((bg_mask) * 255, mode="L").resize((512, 512))
```

```python
# cut = Image.open("data/shadow-gen/stepho_renderings/reflections_4/train/audi-reflection_0001_cut.jpg")
cut_path = image_path.replace(".jpg", "_cut.jpg")
cut = Image.open(cut_path)
cut.resize((512, 512))
```

```python
import cv2
import numpy as np

bg = np.array(image) - np.array(cut)
# Image.fromarray(bg)
# bg
mask = (np.array(cut) < 255).astype(np.uint8)[:, :, 0]
mask = cv2.dilate(mask, np.ones((3, 3), np.uint8), iterations=2)
mask = cv2.erode(mask, np.ones((3, 3), np.uint8), iterations=3)
Image.fromarray(mask * 255, mode="L").resize((512, 512))
# (np.array(cut) < 255).astype(np.uint8)*255
```

```python
car_and_refl_mask = np.logical_or(mask, 1 - bg_mask)
Image.fromarray((car_and_refl_mask * 255).astype(np.uint8), mode="L").resize((512, 512))
```

```python
refl_mask = car_and_refl_mask - mask
Image.fromarray((refl_mask * 255).astype(np.uint8), mode="L").resize((512, 512))
```

```python
opened_mask = cv2.morphologyEx(
    refl_mask, cv2.MORPH_OPEN, np.ones((5, 5), np.uint8), iterations=5
)
Image.fromarray((opened_mask * 255).astype(np.uint8), mode="L").resize((512, 512))
```

```python
closed_mask = cv2.morphologyEx(
    opened_mask, cv2.MORPH_CLOSE, np.ones((5, 5), np.uint8), iterations=8
)
Image.fromarray((closed_mask * 255).astype(np.uint8), mode="L").resize((512, 512))
```
